## Generate Embeddings for Both Documents
Convert the cleaned text into embeddings using the model. Embeddings are numerical vector representations of the text that capture semantic meaning.

````python
# Function to generate embeddings for a given text
def generate_embeddings(text, tokenizer, model):
    # Tokenize the input text
    inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=1024)
    # Generate embeddings using the model
    with torch.no_grad():
        outputs = model(**inputs)
    # Use the mean of the last hidden state as the embedding
    embeddings = outputs.last_hidden_state.mean(dim=1)
    return embeddings

# Generate embeddings for the RFP and service description
rfp_embeddings = generate_embeddings(rfp_text_clean, tokenizer, model)
service_description_embeddings = generate_embeddings(service_description_text_clean, tokenizer, model)
````

````python
import torch
from transformers import AutoTokenizer, AutoModel
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load lm-roberta-base tokenizer and model
model_name = "roberta-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)

# Function to get embeddings using mean pooling
def get_embedding(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)
    with torch.no_grad():
        outputs = model(**inputs)
    return outputs.last_hidden_state.mean(dim=1).numpy()  # Mean pooling


# Compute embedding for service_description_text_clean
service_embedding = get_embedding(service_description_text_clean)

# Reshape to 2D for cosine_similarity
service_embedding = service_embedding.reshape(1, -1)

# Compute embeddings for RFP pages
df_LB["embedding"] = df_LB["rfp_pages_clean"].apply(lambda x: get_embedding(str(x)))

# Convert to numpy arrays
lb_embeddings = np.vstack(df_LB["embedding"].values)

# Compute cosine similarity
similarity_matrix = cosine_similarity(lb_embeddings, service_embedding)

# Create a new DataFrame to store the results
df_similarity_results = pd.DataFrame(data={'Page': df_LB['page_number'], 'Similarity_Score': similarity_matrix[:, 0]})

# Print results
print("Top 10 similarity scores:")
print(df_similarity_results.head(25))

````

````python
from bert_score import score

def compute_bertscore(reference_text, candidate_texts, model_type="roberta-base"):
    """
    Computes BERTScore for a list of candidate texts against a single reference text.
    """
    # Calculate BERTScore for each candidate text
    F1_scores = []
    for candidate_text in candidate_texts:
        P, R, F1 = score([candidate_text], [reference_text], model_type=model_type, lang="de", rescale_with_baseline=True)
        F1_scores.append(F1.mean().item())  # Get the average F1 score for the candidate
    return F1_scores


# Assuming df_LB and service_description_text_clean are defined from previous code
bertscore_values = compute_bertscore(service_description_text_clean, df_LB["rfp_pages_clean"])

# Add the BERTScore values to the DataFrame
df_LB['BERTScore'] = bertscore_values

print("BERTScore F1 for each page:")
print(df_LB[['page_number', 'BERTScore']]) # Print page number and corresponding BERTScore
````

````python

# Load the spaCy german language model
nlp = spacy.load("de_core_news_lg")

def extract_keywords(text, top_n=50):
    doc = nlp(text)
    keywords = []
    for token in doc:
        if token.pos_ in ("NOUN", "PROPN", "ADJ"):
            keywords.append(token.text.lower())
    #  Remove duplicates and return top_n keywords
    return list(dict.fromkeys(keywords))[:top_n]


# Extract keywords from both texts
rfp_keywords = extract_keywords(rfp_text_clean)
service_keywords = extract_keywords(service_description_text_clean)

print("Keywords from RFP:", rfp_keywords)
print("\nKeywords from Service Description:", service_keywords)
````

````python
import spacy
import numpy as np
from scipy.spatial.distance import cdist

# Lade das große deutsche spaCy-Modell
nlp = spacy.load("de_core_news_lg")

def preprocess(text):
    """ Tokenize und entferne Stoppwörter & Satzzeichen. Gibt eine Liste von Tokens zurück. """
    doc = nlp(text)
    tokens = [token for token in doc if not token.is_stop and not token.is_punct and token.has_vector]
    return tokens  # Gibt eine Liste von spaCy-Token-Objekten zurück

def monge_elkan_similarity_optimized(set1, set2):
    """ Berechnet die Monge-Elkan Similarity mit Fehlerbehandlung. """
    if not set1 or not set2:
        return 0  # Falls eine Liste leer ist, Similarity = 0

    # Extrahiere Vektoren
    vectors1 = np.array([token.vector for token in set1 if token.has_vector])
    vectors2 = np.array([token.vector for token in set2 if token.has_vector])

    # Falls keine gültigen Vektoren vorhanden sind, Similarity = 0
    if vectors1.shape[0] == 0 or vectors2.shape[0] == 0:
        return 0

    # Berechne paarweise Kosinus-Distanzen
    distances = cdist(vectors1, vectors2, metric="cosine")

    # Falls alle Distanzen NaN sind, setze Standardwert auf 0
    if np.isnan(distances).all():
        return 0

    # Wähle für jedes Wort in set1 das ähnlichste Wort aus set2
    best_matches = 1 - np.nanmin(distances, axis=1)  # Ähnlichkeit = 1 - Distanz

    return np.nanmean(best_matches) if best_matches.size > 0 else 0  # Falls leer, Similarity = 0

# Beispiel: Vergleich von RFP-Anforderung mit einer Servicebeschreibung
rfp_text = rfp_text_clean
service_description = service_description_text_clean

# Tokenize & Preprocess
rfp_tokens = preprocess(rfp_text)
service_tokens = preprocess(service_description)

# Berechne die optimierte Monge-Elkan Similarity
similarity_score = monge_elkan_similarity_optimized(rfp_tokens, service_tokens)

print(f"Monge-Elkan Similarity: {similarity_score:.4f}")
````

````python


# Load model
nlp = spacy.load("de_core_news_lg")

def extract_named_entities(text):
    """ extract  Named Entities (z. B. products, Services, roles) from the text """
    doc = nlp(text)
    entities = {ent.text for ent in doc.ents if ent.label_ in ["ORG", "PRODUCT", "GPE", "PERSON"]}
    return entities

def compare_entities(rfp_text, service_text):
    """ compare Named Entities between  RFP-pages and service description. """
    rfp_entities = extract_named_entities(rfp_text)
    service_entities = extract_named_entities(service_text)

    common_entities = rfp_entities.intersection(service_entities)
    missing_entities = rfp_entities - service_entities

    return common_entities, missing_entities


service_text = service_description_text_clean


results = []

# go through all Pages
for index, row in df_LB.iterrows():
    page_number = row["page_number"]
    rfp_text = row["rfp_pages_clean"]

    # compare Named Entities
    common, missing = compare_entities(rfp_text, service_text)

    # Save the results in a dataframe list
    results.append({
        "Seite": page_number,
        "Gemeinsame Entitäten": ", ".join(common) if common else "Keine",
        "Fehlende Entitäten": ", ".join(missing) if missing else "Keine"
    })

#convert the results in dataframe
df_results = pd.DataFrame(results)

print(df_results)
````

````python
import pandas as pd
import numpy as np
!python -m nltk.downloader punkt_tab
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
nltk.download('punkt_tab')
import string
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import NMF
from gensim import corpora
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud



# Deutsche Stopwörter laden
stop_words = set(stopwords.words('german'))

def preprocess_text(text):
    # Tokenisierung
    words = nltk.word_tokenize(text)
    # Entferne Stopwörter
    words_filtered = [word for word in words if word.lower() not in stop_words]
    # Verarbeite den Text weiter (z. B. zu einem String zusammenfügen)
    return ' '.join(words_filtered)

# Daten einlesen und Stopwörter entfernen
df_LB["content_clean"] = df_LB["content"].apply(preprocess_text)
service_description_text_clean = preprocess_text(service_description_text_clean)


# TF-IDF Vektorisierung
vectorizer = TfidfVectorizer(max_features=1000)
tfidf_matrix = vectorizer.fit_transform(df_LB["content_clean"].tolist() + [service_description_text_clean])
feature_names = vectorizer.get_feature_names_out()

# Themenmodellierung mit NMF
nmf = NMF(n_components=10, random_state=42)  # 10 Themen
nmf_topics = nmf.fit_transform(tfidf_matrix)

# Extrahiert die wichtigsten Begriffe je Thema
def get_topics(model, feature_names, n_words=10): #10 wörter
    topics = {}
    for topic_idx, topic in enumerate(model.components_):
        topics[f"Topic {topic_idx+1}"] = [feature_names[i] for i in topic.argsort()[:-n_words - 1:-1]]
    return topics

nmf_topics_dict = get_topics(nmf, feature_names)

# Create a heatmap of the topic distribution
plt.figure(figsize=(10, 6))
sns.heatmap(nmf_topics, annot=False, cmap="viridis", fmt=".2f")
plt.xlabel("Topics")
plt.ylabel("Documents")
plt.title("Topic Distribution across Documents (NMF)")
plt.show()


# Function to display top words per topic in a more readable format
def display_topics(model, feature_names, n_top_words):
    for topic_idx, topic in enumerate(model.components_):
        print(f"Topic {topic_idx +1}:")
        print(" ".join([feature_names[i]
                        for i in topic.argsort()[:-n_top_words - 1:-1]]))

# Example usage (adjust n_top_words as needed):
display_topics(nmf, feature_names, 10)

def create_wordcloud(topic_words):
    """Creates a word cloud from a list of words."""
    wordcloud = WordCloud(width=800, height=400, background_color='white').generate(" ".join(topic_words))
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis("off")
    plt.show()

# Create word clouds for each topic
for topic, words in nmf_topics_dict.items():
    print(f"Word cloud for: {topic}")
    create_wordcloud(words)

````

# Transformer-basierte Frage-Antwort-Modelle (QA)

✅ Identifizierung relevanter Passagen: Statt nur Zahlenwerten zeigt das Modell auch Textstellen, die als Antwort infrage kommen.

````python
import pandas as pd
from transformers import pipeline

# Lade ein vortrainiertes QA-Modell
qa_pipeline = pipeline("question-answering", model="deepset/gelectra-large-generator")


#Beantwortet eine Liste von Fragen über das gesamte RFP-Dokument basierend auf Seiten. """
def answer_rfp_questions(df_LB, questions):

    results = []

    # Iteriere durch jede Frage
    for question in questions:
        best_answer = None
        best_score = 0.0
        top_answers = []

        # Iteriere durch jede Seite im DataFrame
        for _, row in df_LB.iterrows():
            page_num = row['page_number']
            chunk = row['rfp_pages_clean']

            try:
                # Wende das Modell auf jede Seite an
                response = qa_pipeline(question=question, context=chunk)

                # Vergleiche den Score und füge den besten Antwort-Score hinzu
                if response["score"] > best_score:
                    best_answer = response["answer"]
                    best_score = response["score"]
                    top_answers.append({"Seite": page_num, "Antwort": best_answer, "Wahrscheinlichkeit": best_score})

            except Exception as e:
                print(f"Fehler bei Frage '{question}' auf Seite {page_num}: {e}")

        # Wähle die beste Antwort aus den gesammelten Ergebnissen
        if top_answers:
            best_answer = max(top_answers, key=lambda x: x['Wahrscheinlichkeit'])
            results.append({
                "Frage": question,
                "Antwort": best_answer["Antwort"],
                "Wahrscheinlichkeit": best_answer["Wahrscheinlichkeit"],
                "Seite": best_answer["Seite"]
            })

    return results

# Beispiel-RFP-Fragen
rfp_questions = [
    "Wie viele Benutzer sind in dem System registriert?",
    "An wie vielen Standorten wird das System eingesetzt?",
    "Welche Microsoft 365-Dienste werden für das Projekt benötigt?",
    "Ist eine Migration von bestehenden Systemen erforderlich?",
    "Werden Shared-Arbeitsplätze im Rahmen des Projekts benötigt?"
]


# Frage-Antwort-Prozess ausführen
qa_results = answer_rfp_questions(df_LB, rfp_questions)

# Ergebnisse anzeigen
df_results = pd.DataFrame(qa_results)
print(df_results)
````
